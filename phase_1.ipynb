```python
# Now checking which is faster 
#1 list multiplication in py or numpy arr multiplication
import time as t
start=t.time()
list1_py=[i*2 for i in range(100)]
print(list1_py)
print("time taken to perform operations: ",t.time()-start)


start1=t.time()
arr=np.arange(100)*2
print(arr)
print("time taken to perform operations: ",t.time()-start1)

```

### Properties of the arrays

```python
#shape and dimension and size
matrix=np.array([[1,2,3,4],[4,3,6,7],[0,8,9,1]])
print("shape ",matrix.shape)
print("dimension ",matrix.ndim)
print("size ",matrix.size)

print("data type ",matrix.dtype)
```

### Reshape the arrays

```python
arr=np.arange(12)
print("originl array ",arr)

reshaped_ar=arr.reshape((4,3))
print("\n reshapeed array ",reshaped_ar)

print()
flat_arr=reshaped_ar.flatten() # make a copy of the array in 1d 
print("flatten array ",flat_arr)

print()
raveled_ar=reshaped_ar.ravel() # dont make any copy 
print("ravel array ",raveled_ar)
print()

raveled_ar[5]=89 # i have make changes in the raveled but it also changes the original one 
print(reshaped_ar)


```

```python
# transpose of the matrix
# matrix.transpose     or  matrix.T

matrix=np.array([[4,9,2],
                 [3,8,1],
                 [5,6,0]])
trans_mat=matrix.transpose()
print(trans_mat) 

print()

mat2=np.array([[4,9,2],
                 [3,8,1],
                 [5,6,0]])

mat3=mat2 + matrix
print(mat3)
print()
tran_mat3=mat3.T
print(tran_mat3)

# if you want to make sum of all the elements of the arr
print()
sum_of_all= np.sum([matrix])
print(sum_of_all)

```
