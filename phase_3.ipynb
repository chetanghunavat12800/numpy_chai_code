```python
# data structure: [restaurant_id,2021,2022,2023,2024]
sales_data=np.array([
    [1,150000,180000,220000,250000] ,#paradise
    [2,120000,140000,160000,190000],#beijing
    [3,200000,230000,260000,300000],# pizza hub
    [4,180000,210000,240000,270000], #burger point
    [5,160000,185000,205000,230000] #chai point
])

print("sale datashape: ",sales_data.shape) # to know the shape of the data
print("====Zomato sales analysis====")
print("sales data for the 1st 3 restaurants : ")
print(sales_data[np.arange(3),:])

print()

print(sales_data[0:3]) # sales_data[:3]
print()
print(sales_data[:,:3])
print()
print(sales_data[:,1:]) # this print only sales not the restaurant id
print(sales_data[:3,1:])

# to find the total sales of each restaurant in four years

# for i in range(5):
#     sum_is=np.sum(sales_data[i,1:])
#     print(f"the sum of the {i+1} id restaurant :",sum_is)
print()
print(np.sum(sales_data[:,1:],axis=1))

#total sales per year ...sum of all sales in one year
print()
print(np.sum(sales_data[:,1:],axis=0))

# min.& max sales per restaurant 
print("min sales per restaurant: ",np.min(sales_data[:,1:],axis=1))
print()
print('max sales per restaurant: ',np.max(sales_data[:,1:],axis=1))


# average sale per year of each restaurant 
print()
print("average sale per year of each: ",np.mean(sales_data[:,1:],axis=1))

```
